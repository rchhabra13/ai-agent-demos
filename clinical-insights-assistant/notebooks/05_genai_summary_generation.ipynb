# Notebook 05: GenAI Summary Generation

Use LLM-based summarization for doctor notes, regulatory summaries, and cohort narratives.

> **Note:** This notebook uses a mock LLM for self-contained operation. Swap `MockLLM` for `OpenAI` calls in production.

```python
import pandas as pd
import sys, os

sys.path.insert(0, os.path.join(os.getcwd(), '..'))
from src.data_loader import load_data
from src.genai_interface import (
    summarize_doctor_notes, generate_regulatory_summary,
    generate_recommendations, generate_cohort_narrative,
    summarize_scenario_results, get_llm
)
from src.cohort_analysis import run_full_cohort_analysis
from src.issue_detection import run_all_detections
from src.scenario_simulation import simulate_scenario

df = load_data('../data/clinical_trial_data.csv')
llm = get_llm()
print(f'Data loaded: {len(df)} records')
print(f'LLM backend: {llm.model_name}')
```

## 1. Doctor Notes Summarization

Extract key themes, adverse events, and anomalies from doctor feedback.

```python
notes_summary = summarize_doctor_notes(df)
print(notes_summary)
```

## 2. Regulatory Summary Generation

Generate a 3-paragraph FDA-style regulatory summary aligned with submission expectations.

```python
# First run supporting analyses
detection_results = run_all_detections(df)
cohort_results = run_full_cohort_analysis(df)

reg_summary = generate_regulatory_summary(df, cohort_results, detection_results)
print(reg_summary)
```

## 3. Cohort Comparison Narrative

```python
cohort_narrative = generate_cohort_narrative(cohort_results)
print(cohort_narrative)
```

## 4. Scenario Simulation Summary

```python
scenario_result = simulate_scenario(df, 'sensitivity_analysis')
scenario_summary = summarize_scenario_results(scenario_result)
print(scenario_summary)
```

## 5. Recommended Next Steps

```python
context = {
    'issues_found': detection_results['summary']['total_issues_found'],
    'total_patients': detection_results['summary']['total_patients'],
    'cohort_significant': cohort_results['outcome_comparison']['significant_at_05'],
    'optimal_dosage': scenario_result['optimal_dosage']
}
recommendations = generate_recommendations(context)
print(recommendations)
```

## 6. Combined Report Generation

```python
print('=' * 60)
print('CLINICAL INSIGHTS ASSISTANT - COMPREHENSIVE REPORT')
print('=' * 60)

print('\n--- SECTION 1: Doctor Notes Analysis ---')
print(notes_summary[:500])

print('\n--- SECTION 2: Issue Detection Summary ---')
print(f'Total issues: {detection_results["summary"]["total_issues_found"]}')
for key in ['non_compliance', 'adverse_events', 'outcome_anomalies']:
    print(f'  {key}: {detection_results[key]["count"]} issues')

print('\n--- SECTION 3: Cohort Comparison ---')
print(cohort_narrative[:500])

print('\n--- SECTION 4: Regulatory Summary ---')
print(reg_summary[:500])

print('\n--- SECTION 5: Recommendations ---')
print(recommendations[:500])

print(f'\nLLM calls made: {llm.call_count}')
```

- Regulatory summaries follow FDA-style formatting (executive summary, efficacy, safety)
