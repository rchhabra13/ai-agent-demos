# Notebook 06: Agentic AI Loop

Demonstrate the autonomous agent that reasons across tools, memory, and analysis modules.

The agent uses a **ReAct-style** (Reason + Act) loop to:
1. Think about what analysis to perform next
2. Execute the appropriate tool
3. Observe and record the results
4. Repeat until the task is complete

```python
import pandas as pd
import sys, os

sys.path.insert(0, os.path.join(os.getcwd(), '..'))
from src.data_loader import load_data
from src.agent.agent_core import ClinicalInsightsAgent

df = load_data('../data/clinical_trial_data.csv')
print(f'Data loaded: {len(df)} records')
```

## 1. Initialize the Agent

```python
agent = ClinicalInsightsAgent(df, verbose=True)
print(f'Agent initialized with {len(df)} records')
print(f'Available tools: {list(agent.AVAILABLE_TOOLS.keys())}')
```

## 2. Run Comprehensive Analysis

The agent will autonomously decide which analyses to run and in what order.

```python
report = agent.run('Perform a comprehensive analysis of the clinical trial data.')

print(f'\n{"="*60}')
print('AGENT EXECUTION SUMMARY')
print(f'{"="*60}')
print(f'Session ID: {report["session_id"]}')
print(f'Total reasoning steps: {report["summary"]["num_reasoning_steps"]}')
print(f'Analysis results stored: {report["summary"]["num_results"]}')
print(f'Result keys: {report["summary"]["result_keys"]}')
```

## 3. Inspect Reasoning Trace

```python
trace = report['reasoning_trace']
for step in trace:
    print(f'\n--- Step {step["step"]} ---')
    print(f'  THINK: {step["thought"]}')
    print(f'  ACT: {step["action"]}')
    print(f'  OBSERVE: {step["observation"][:200]}...' if len(step['observation']) > 200 else f'  OBSERVE: {step["observation"]}')
```

## 4. Interactive Q&A with the Agent

```python
# Ask specific questions
questions = [
    'Tell me about patient P042',
    'What are the adverse event patterns?',
    'Compare the cohorts for me',
    'What dosage would you recommend?',
    'Generate a regulatory summary',
]

for q in questions:
    print(f'\n{"="*50}')
    print(f'Q: {q}')
    print(f'{"="*50}')
    answer = agent.ask(q)
    print(answer[:500] if len(answer) > 500 else answer)
```

## 5. Agent Memory Inspection

```python
# Inspect what the agent remembers
memory = agent.memory

print('=== CONVERSATION HISTORY ===')
for msg in memory.conversation.get_history()[-5:]:
    print(f'  [{msg["role"]}]: {msg["content"][:100]}...')

print(f'\n=== ANALYSIS MEMORY ===')
print(f'Results stored: {len(memory.analysis.results)}')
for key in memory.analysis.results:
    print(f'  - {key}')

print(f'\nReasoning steps: {len(memory.analysis.reasoning_trace)}')
```

## 6. Export Agent Report

```python
import json

# Prepare exportable report
export_report = {
    'session_id': report['session_id'],
    'reasoning_trace': report['reasoning_trace'],
    'summary': report['summary'],
}

# Save to file
with open('../data/agent_report.json', 'w') as f:
    json.dump(export_report, f, indent=2, default=str)

print('Agent report exported to data/agent_report.json')
print(f'Report contains {len(export_report["reasoning_trace"])} reasoning steps')
```

## Key Takeaways

- The agent autonomously plans and executes a multi-step analysis pipeline
- ReAct-style reasoning provides transparency into decision-making
- Session memory enables context-aware follow-up queries
- The agent can be extended with additional tools and more sophisticated planning
- All results are traceable through the reasoning trace
